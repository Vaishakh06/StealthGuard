```python
# =============================================================================
# SECTION 3 — VERIFY MITIGATION INPUTS
# CELL 2 — Verify Decision-State and Security-Severity Distributions
# =============================================================================

print("=" * 80)
print("VERIFYING NOTEBOOK 06 DECISION DISTRIBUTIONS")
print("=" * 80)

# ---------------------------------------------------------------------------
# Expected distributions from the verified Notebook 06 summary
# ---------------------------------------------------------------------------

EXPECTED_DECISION_STATES = {
    "NORMAL": 270_361,
    "SUSPICIOUS": 37_300,
    "HIGH_EVIDENCE": 30_784,
}

EXPECTED_SECURITY_SEVERITIES = {
    "LOW": 261_669,
    "MEDIUM": 36_607,
    "HIGH": 30_736,
    "NORMAL": 8_692,
    "MONITOR": 693,
    "INVESTIGATE": 48,
}

# ---------------------------------------------------------------------------
# Actual distributions
# ---------------------------------------------------------------------------

actual_decision_states = (
    mitigation_df["Decision_State"]
    .value_counts()
    .to_dict()
)

actual_security_severities = (
    mitigation_df["Security_Severity"]
    .value_counts()
    .to_dict()
)

# ---------------------------------------------------------------------------
# Decision-state verification
# ---------------------------------------------------------------------------

print("\n1. Decision-State Counts")
print("-" * 80)

decision_state_pass = True

for state, expected_count in EXPECTED_DECISION_STATES.items():
    actual_count = actual_decision_states.get(state, 0)
    passed = actual_count == expected_count

    print(
        f"{state:<15} "
        f"Expected: {expected_count:>8,} | "
        f"Actual: {actual_count:>8,} | "
        f"PASS: {passed}"
    )

    if not passed:
        decision_state_pass = False

unexpected_decision_states = set(actual_decision_states) - set(
    EXPECTED_DECISION_STATES
)

if unexpected_decision_states:
    print(
        f"\nUnexpected decision states: "
        f"{sorted(unexpected_decision_states)}"
    )
    decision_state_pass = False

# ---------------------------------------------------------------------------
# Security-severity verification
# ---------------------------------------------------------------------------

print("\n2. Security-Severity Counts")
print("-" * 80)

security_severity_pass = True

for severity, expected_count in EXPECTED_SECURITY_SEVERITIES.items():
    actual_count = actual_security_severities.get(severity, 0)
    passed = actual_count == expected_count

    print(
        f"{severity:<15} "
        f"Expected: {expected_count:>8,} | "
        f"Actual: {actual_count:>8,} | "
        f"PASS: {passed}"
    )

    if not passed:
        security_severity_pass = False

unexpected_security_severities = set(actual_security_severities) - set(
    EXPECTED_SECURITY_SEVERITIES
)

if unexpected_security_severities:
    print(
        f"\nUnexpected security severities: "
        f"{sorted(unexpected_security_severities)}"
    )
    security_severity_pass = False

# ---------------------------------------------------------------------------
# Overall verification
# ---------------------------------------------------------------------------

distribution_verification_pass = (
    decision_state_pass
    and security_severity_pass
)

print("\n" + "=" * 80)
print(
    f"DECISION DISTRIBUTION VERIFICATION : "
    f"{decision_state_pass}"
)
print(
    f"SEVERITY DISTRIBUTION VERIFICATION : "
    f"{security_severity_pass}"
)
print("-" * 80)
print(
    f"OVERALL DISTRIBUTION VERIFICATION  : "
    f"{distribution_verification_pass}"
)
print("=" * 80)

if not distribution_verification_pass:
    raise ValueError(
        "Notebook 06 decision/severity distributions do not match "
        "the verified mitigation summary."
    )
```

# Section 4

```python
# =============================================================================
# SECTION 4 — MITIGATION DECISION ENGINE
# CELL 1 — Define Severity-Based Mitigation Policy
# =============================================================================

print("=" * 80)
print("MITIGATION DECISION ENGINE")
print("=" * 80)

# ---------------------------------------------------------------------------
# Simulation-only mitigation policy
# ---------------------------------------------------------------------------
#
# The mitigation engine does NOT perform real network actions.
# It only assigns the mitigation action that would be recommended
# for each security severity.
#
# Severity meaning inherited from Notebook 06:
#
# NORMAL       -> No security action required
# MONITOR      -> Monitor traffic / retain for observation
# INVESTIGATE  -> Security investigation required
# LOW          -> Low-priority DDoS mitigation
# MEDIUM       -> Rate limiting
# HIGH         -> IP blocking + rate limiting
#
# ---------------------------------------------------------------------------

MITIGATION_POLICY = {
    "NORMAL": {
        "action": "NO_ACTION",
        "priority": "NONE",
        "simulation": "No mitigation required",
    },

    "MONITOR": {
        "action": "MONITOR",
        "priority": "LOW",
        "simulation": "Traffic monitored for continued suspicious activity",
    },

    "INVESTIGATE": {
        "action": "INVESTIGATE",
        "priority": "HIGH",
        "simulation": "Security investigation triggered",
    },

    "LOW": {
        "action": "ALERT",
        "priority": "LOW",
        "simulation": "Low-priority DDoS alert generated",
    },

    "MEDIUM": {
        "action": "RATE_LIMIT",
        "priority": "MEDIUM",
        "simulation": "Rate limiting simulated",
    },

    "HIGH": {
        "action": "BLOCK_AND_RATE_LIMIT",
        "priority": "CRITICAL",
        "simulation": "IP blocking and rate limiting simulated",
    },
}

# ---------------------------------------------------------------------------
# Policy verification
# ---------------------------------------------------------------------------

expected_severities = set(EXPECTED_SECURITY_SEVERITIES.keys())
policy_severities = set(MITIGATION_POLICY.keys())

policy_complete = expected_severities == policy_severities

print("\nConfigured mitigation policies:")

for severity, policy in MITIGATION_POLICY.items():
    print(
        f"\n{severity}"
        f"\n  Action     : {policy['action']}"
        f"\n  Priority   : {policy['priority']}"
        f"\n  Simulation : {policy['simulation']}"
    )

print("\n" + "-" * 80)
print(f"Policy covers all Notebook 06 severities : {policy_complete}")
print("-" * 80)

if not policy_complete:
    raise ValueError(
        "Mitigation policy does not cover every security severity "
        "produced by Notebook 06."
    )
```

```python
# =============================================================================
# SECTION 4 — MITIGATION DECISION ENGINE
# CELL 2 — Finalize Severity-Based Mitigation Policy
# =============================================================================

print("=" * 80)
print("FINALIZING SEVERITY-BASED MITIGATION POLICY")
print("=" * 80)

# ---------------------------------------------------------------------------
# Final mitigation policy
# ---------------------------------------------------------------------------
#
# IMPORTANT:
# The mitigation engine is simulation-based.
#
# It determines what action SHOULD be taken for each sample but does not
# execute real firewall, router, IP-blocking, or rate-limiting commands.
#
# LOW:
#   Predicted DDoS + NORMAL AnomalyFlow evidence.
#   Traffic is NOT considered safe.
#   We alert and log rather than automatically block.
#
# MEDIUM:
#   Predicted DDoS + SUSPICIOUS evidence.
#   Rate limiting is simulated.
#
# HIGH:
#   Predicted DDoS + HIGH_EVIDENCE.
#   Blocking and rate limiting are simulated.
# ---------------------------------------------------------------------------

MITIGATION_POLICY = {
    "NORMAL": {
        "action": "NO_ACTION",
        "priority": "NONE",
        "simulation": "No mitigation required",
    },

    "MONITOR": {
        "action": "MONITOR",
        "priority": "LOW",
        "simulation": "Traffic monitored for continued suspicious activity",
    },

    "INVESTIGATE": {
        "action": "INVESTIGATE",
        "priority": "HIGH",
        "simulation": "Security investigation triggered",
    },

    "LOW": {
        "action": "ALERT_AND_LOG",
        "priority": "LOW",
        "simulation": "DDoS alert generated and traffic event logged",
    },

    "MEDIUM": {
        "action": "RATE_LIMIT",
        "priority": "MEDIUM",
        "simulation": "Traffic rate limiting simulated",
    },

    "HIGH": {
        "action": "BLOCK_AND_RATE_LIMIT",
        "priority": "CRITICAL",
        "simulation": "Traffic blocking and rate limiting simulated",
    },
}

# ---------------------------------------------------------------------------
# Verify policy completeness
# ---------------------------------------------------------------------------

expected_severities = set(EXPECTED_SECURITY_SEVERITIES.keys())
policy_severities = set(MITIGATION_POLICY.keys())

policy_complete = expected_severities == policy_severities

print("\nFinal mitigation policy:")

for severity, policy in MITIGATION_POLICY.items():
    print(
        f"\n{severity}"
        f"\n  Action     : {policy['action']}"
        f"\n  Priority   : {policy['priority']}"
        f"\n  Simulation : {policy['simulation']}"
    )

print("\n" + "-" * 80)
print(f"Policy covers all Notebook 06 severities : {policy_complete}")
print("-" * 80)

if not policy_complete:
    raise ValueError(
        "Final mitigation policy does not cover every "
        "Notebook 06 security severity."
    )
```

```python
# =============================================================================
# SECTION 4 — MITIGATION DECISION ENGINE
# CELL 3 — Apply Final Mitigation Policy
# =============================================================================

print("=" * 80)
print("APPLYING FINAL MITIGATION POLICY")
print("=" * 80)

# ---------------------------------------------------------------------------
# Map security severity → mitigation action
# ---------------------------------------------------------------------------

severity_to_action = {
    severity: policy["action"]
    for severity, policy in MITIGATION_POLICY.items()
}

severity_to_priority = {
    severity: policy["priority"]
    for severity, policy in MITIGATION_POLICY.items()
}

severity_to_simulation = {
    severity: policy["simulation"]
    for severity, policy in MITIGATION_POLICY.items()
}

# ---------------------------------------------------------------------------
# Apply mappings
# ---------------------------------------------------------------------------

mitigation_df["Mitigation_Action"] = (
    mitigation_df["Security_Severity"]
    .map(severity_to_action)
)

mitigation_df["Mitigation_Priority"] = (
    mitigation_df["Security_Severity"]
    .map(severity_to_priority)
)

mitigation_df["Mitigation_Simulation"] = (
    mitigation_df["Security_Severity"]
    .map(severity_to_simulation)
)

# ---------------------------------------------------------------------------
# Verify complete mapping
# ---------------------------------------------------------------------------

missing_actions = mitigation_df["Mitigation_Action"].isna().sum()
missing_priorities = mitigation_df["Mitigation_Priority"].isna().sum()
missing_simulations = mitigation_df["Mitigation_Simulation"].isna().sum()

mapping_pass = (
    missing_actions == 0
    and missing_priorities == 0
    and missing_simulations == 0
)

print("\nMapping verification:")
print(f"Missing mitigation actions      : {missing_actions}")
print(f"Missing mitigation priorities   : {missing_priorities}")
print(f"Missing simulation descriptions : {missing_simulations}")
print(f"PASS                             : {mapping_pass}")

# ---------------------------------------------------------------------------
# Mitigation action distribution
# ---------------------------------------------------------------------------

print("\n" + "-" * 80)
print("MITIGATION ACTION DISTRIBUTION")
print("-" * 80)

action_counts = (
    mitigation_df["Mitigation_Action"]
    .value_counts()
)

for action, count in action_counts.items():
    percentage = (count / len(mitigation_df)) * 100

    print(
        f"{action:<24}"
        f"{count:>10,} "
        f"({percentage:>6.2f}%)"
    )

# ---------------------------------------------------------------------------
# Cross-check: severity → action
# ---------------------------------------------------------------------------

print("\n" + "-" * 80)
print("SEVERITY → MITIGATION ACTION")
print("-" * 80)

severity_action_table = (
    mitigation_df[
        ["Security_Severity", "Mitigation_Action"]
    ]
    .drop_duplicates()
    .sort_values("Security_Severity")
)

display(severity_action_table)

# ---------------------------------------------------------------------------
# Preview final mitigation decisions
# ---------------------------------------------------------------------------

print("\n" + "-" * 80)
print("SAMPLE MITIGATION DECISIONS")
print("-" * 80)

display(
    mitigation_df[
        [
            "Test_Index",
            "Predicted_Label",
            "AnomalyFlow_Score",
            "Decision_State",
            "Security_Severity",
            "Mitigation_Action",
            "Mitigation_Priority",
        ]
    ].head(10)
)

# ---------------------------------------------------------------------------
# Final validation
# ---------------------------------------------------------------------------

if not mapping_pass:
    raise ValueError(
        "One or more samples could not be assigned "
        "a mitigation decision."
    )

print("\n" + "=" * 80)
print("MITIGATION POLICY APPLICATION : PASS")
print("=" * 80)
```

# Section 5

```python
# =============================================================================
# SECTION 5 — ALERT GENERATION
# CELL 1 — Define Alert Policy
# =============================================================================

print("=" * 80)
print("ALERT GENERATION")
print("=" * 80)

# ---------------------------------------------------------------------------
# Alert policy
# ---------------------------------------------------------------------------
#
# Alerts are generated from the final mitigation decision.
#
# NO_ACTION:
#   No security alert.
#
# MONITOR:
#   Monitoring alert.
#
# INVESTIGATE:
#   Investigation alert.
#
# ALERT_AND_LOG:
#   DDoS detection alert.
#
# RATE_LIMIT:
#   Rate-limit mitigation alert.
#
# BLOCK_AND_RATE_LIMIT:
#   Critical blocking alert.
# ---------------------------------------------------------------------------

ALERT_POLICY = {
    "NO_ACTION": {
        "alert_generated": False,
        "alert_level": "NONE",
        "alert_type": "NONE",
        "message": "No security action required.",
    },

    "MONITOR": {
        "alert_generated": True,
        "alert_level": "LOW",
        "alert_type": "MONITORING",
        "message": "Suspicious traffic requires continued monitoring.",
    },

    "INVESTIGATE": {
        "alert_generated": True,
        "alert_level": "HIGH",
        "alert_type": "INVESTIGATION",
        "message": "Non-DDoS traffic shows high security evidence and requires investigation.",
    },

    "ALERT_AND_LOG": {
        "alert_generated": True,
        "alert_level": "LOW",
        "alert_type": "DDOS_DETECTED",
        "message": "DDoS traffic detected with low AnomalyFlow evidence; event logged.",
    },

    "RATE_LIMIT": {
        "alert_generated": True,
        "alert_level": "MEDIUM",
        "alert_type": "RATE_LIMIT_TRIGGERED",
        "message": "DDoS traffic detected with suspicious evidence; rate limiting simulated.",
    },

    "BLOCK_AND_RATE_LIMIT": {
        "alert_generated": True,
        "alert_level": "CRITICAL",
        "alert_type": "BLOCK_TRIGGERED",
        "message": "High-evidence DDoS traffic detected; blocking and rate limiting simulated.",
    },
}

# ---------------------------------------------------------------------------
# Verify alert policy completeness
# ---------------------------------------------------------------------------

expected_actions = set(MITIGATION_POLICY.keys())

# The policy keys are severities, so obtain all actual mitigation actions.
expected_actions = set(MITIGATION_POLICY[s]["action"] for s in MITIGATION_POLICY)

alert_policy_actions = set(ALERT_POLICY.keys())

alert_policy_complete = expected_actions == alert_policy_actions

print("\nConfigured alert policies:")

for action, policy in ALERT_POLICY.items():
    print(
        f"\n{action}"
        f"\n  Alert Generated : {policy['alert_generated']}"
        f"\n  Alert Level     : {policy['alert_level']}"
        f"\n  Alert Type      : {policy['alert_type']}"
        f"\n  Message         : {policy['message']}"
    )

print("\n" + "-" * 80)
print(f"Alert policy covers all mitigation actions : {alert_policy_complete}")
print("-" * 80)

if not alert_policy_complete:
    raise ValueError(
        "Alert policy does not cover every mitigation action."
    )
```

```python
# =============================================================================
# SECTION 5 — ALERT GENERATION
# CELL 2 — Generate Sample-Level Security Alerts
# =============================================================================

print("=" * 80)
print("GENERATING SAMPLE-LEVEL SECURITY ALERTS")
print("=" * 80)

# ---------------------------------------------------------------------------
# Create lookup dictionaries
# ---------------------------------------------------------------------------

action_to_alert_generated = {
    action: policy["alert_generated"]
    for action, policy in ALERT_POLICY.items()
}

action_to_alert_level = {
    action: policy["alert_level"]
    for action, policy in ALERT_POLICY.items()
}

action_to_alert_type = {
    action: policy["alert_type"]
    for action, policy in ALERT_POLICY.items()
}

action_to_alert_message = {
    action: policy["message"]
    for action, policy in ALERT_POLICY.items()
}

# ---------------------------------------------------------------------------
# Apply alert policy
# ---------------------------------------------------------------------------

mitigation_df["Alert_Generated"] = (
    mitigation_df["Mitigation_Action"]
    .map(action_to_alert_generated)
)

mitigation_df["Alert_Level"] = (
    mitigation_df["Mitigation_Action"]
    .map(action_to_alert_level)
)

mitigation_df["Alert_Type"] = (
    mitigation_df["Mitigation_Action"]
    .map(action_to_alert_type)
)

mitigation_df["Alert_Message"] = (
    mitigation_df["Mitigation_Action"]
    .map(action_to_alert_message)
)

# ---------------------------------------------------------------------------
# Verify complete alert mapping
# ---------------------------------------------------------------------------

missing_alert_flags = mitigation_df["Alert_Generated"].isna().sum()
missing_alert_levels = mitigation_df["Alert_Level"].isna().sum()
missing_alert_types = mitigation_df["Alert_Type"].isna().sum()
missing_alert_messages = mitigation_df["Alert_Message"].isna().sum()

alert_mapping_pass = (
    missing_alert_flags == 0
    and missing_alert_levels == 0
    and missing_alert_types == 0
    and missing_alert_messages == 0
)

print("\nAlert mapping verification:")
print(f"Missing alert flags     : {missing_alert_flags}")
print(f"Missing alert levels    : {missing_alert_levels}")
print(f"Missing alert types     : {missing_alert_types}")
print(f"Missing alert messages  : {missing_alert_messages}")
print(f"PASS                    : {alert_mapping_pass}")

# ---------------------------------------------------------------------------
# Alert generation distribution
# ---------------------------------------------------------------------------

print("\n" + "-" * 80)
print("ALERT GENERATION DISTRIBUTION")
print("-" * 80)

alert_counts = (
    mitigation_df["Alert_Generated"]
    .value_counts()
    .sort_index()
)

for generated, count in alert_counts.items():
    percentage = (count / len(mitigation_df)) * 100

    print(
        f"Alert Generated = {generated:<5} "
        f"{count:>10,} "
        f"({percentage:>6.2f}%)"
    )

# ---------------------------------------------------------------------------
# Alert type distribution
# ---------------------------------------------------------------------------

print("\n" + "-" * 80)
print("ALERT TYPE DISTRIBUTION")
print("-" * 80)

alert_type_counts = (
    mitigation_df["Alert_Type"]
    .value_counts()
)

for alert_type, count in alert_type_counts.items():
    percentage = (count / len(mitigation_df)) * 100

    print(
        f"{alert_type:<24}"
        f"{count:>10,} "
        f"({percentage:>6.2f}%)"
    )

# ---------------------------------------------------------------------------
# Alert level distribution
# ---------------------------------------------------------------------------

print("\n" + "-" * 80)
print("ALERT LEVEL DISTRIBUTION")
print("-" * 80)

alert_level_counts = (
    mitigation_df["Alert_Level"]
    .value_counts()
)

for level, count in alert_level_counts.items():
    percentage = (count / len(mitigation_df)) * 100

    print(
        f"{level:<12}"
        f"{count:>10,} "
        f"({percentage:>6.2f}%)"
    )

# ---------------------------------------------------------------------------
# Preview generated alerts
# ---------------------------------------------------------------------------

print("\n" + "-" * 80)
print("SAMPLE GENERATED ALERTS")
print("-" * 80)

display(
    mitigation_df[
        [
            "Test_Index",
            "Predicted_Label",
            "Decision_State",
            "Security_Severity",
            "Mitigation_Action",
            "Alert_Generated",
            "Alert_Level",
            "Alert_Type",
        ]
    ].head(10)
)

# ---------------------------------------------------------------------------
# Final validation
# ---------------------------------------------------------------------------

if not alert_mapping_pass:
    raise ValueError(
        "One or more samples could not be assigned "
        "a complete alert decision."
    )

print("\n" + "=" * 80)
print("SAMPLE-LEVEL ALERT GENERATION : PASS")
print("=" * 80)
```

```python
# =============================================================================
# SECTION 6 — MITIGATION SIMULATION
# CELL 1 — Generate Mitigation Simulation Flags
# =============================================================================

print("=" * 80)
print("MITIGATION SIMULATION")
print("=" * 80)

# ---------------------------------------------------------------------------
# Simulation flags
# ---------------------------------------------------------------------------
#
# These flags represent what the mitigation engine WOULD do.
# No real network action is executed.
# ---------------------------------------------------------------------------

mitigation_df["Simulated_Block"] = (
    mitigation_df["Mitigation_Action"]
    == "BLOCK_AND_RATE_LIMIT"
)

mitigation_df["Simulated_Rate_Limit"] = (
    mitigation_df["Mitigation_Action"].isin(
        [
            "RATE_LIMIT",
            "BLOCK_AND_RATE_LIMIT",
        ]
    )
)

mitigation_df["Simulated_Monitoring"] = (
    mitigation_df["Mitigation_Action"]
    == "MONITOR"
)

mitigation_df["Simulated_Investigation"] = (
    mitigation_df["Mitigation_Action"]
    == "INVESTIGATE"
)

# ---------------------------------------------------------------------------
# Verify flag integrity
# ---------------------------------------------------------------------------

expected_block_count = (
    mitigation_df["Security_Severity"] == "HIGH"
).sum()

expected_rate_limit_count = (
    mitigation_df["Security_Severity"].isin(
        ["MEDIUM", "HIGH"]
    )
).sum()

expected_monitor_count = (
    mitigation_df["Security_Severity"] == "MONITOR"
).sum()

expected_investigation_count = (
    mitigation_df["Security_Severity"] == "INVESTIGATE"
).sum()

actual_block_count = mitigation_df["Simulated_Block"].sum()
actual_rate_limit_count = mitigation_df["Simulated_Rate_Limit"].sum()
actual_monitor_count = mitigation_df["Simulated_Monitoring"].sum()
actual_investigation_count = mitigation_df["Simulated_Investigation"].sum()

simulation_pass = (
    actual_block_count == expected_block_count
    and actual_rate_limit_count == expected_rate_limit_count
    and actual_monitor_count == expected_monitor_count
    and actual_investigation_count == expected_investigation_count
)

print("\nSimulation counts:")
print(
    f"Simulated blocking       : "
    f"{actual_block_count:,}"
)

print(
    f"Simulated rate limiting : "
    f"{actual_rate_limit_count:,}"
)

print(
    f"Simulated monitoring    : "
    f"{actual_monitor_count:,}"
)

print(
    f"Simulated investigation : "
    f"{actual_investigation_count:,}"
)

print("\n" + "-" * 80)
print(f"MITIGATION FLAG VERIFICATION : {simulation_pass}")
print("-" * 80)

if not simulation_pass:
    raise ValueError(
        "Mitigation simulation flags do not match "
        "the defined mitigation policy."
    )

# ---------------------------------------------------------------------------
# Display action matrix
# ---------------------------------------------------------------------------

print("\n" + "-" * 80)
print("MITIGATION ACTION MATRIX")
print("-" * 80)

action_matrix = (
    mitigation_df[
        [
            "Security_Severity",
            "Mitigation_Action",
            "Simulated_Block",
            "Simulated_Rate_Limit",
            "Simulated_Monitoring",
            "Simulated_Investigation",
        ]
    ]
    .drop_duplicates()
    .sort_values("Security_Severity")
)

display(action_matrix)

print("\n" + "=" * 80)
print("MITIGATION SIMULATION FLAGS : PASS")
print("=" * 80)
```

```python
# =============================================================================
# SECTION 6 — MITIGATION SIMULATION
# CELL 2 — Generate Mitigation Outcome
# =============================================================================

print("=" * 80)
print("GENERATING MITIGATION OUTCOMES")
print("=" * 80)

# ---------------------------------------------------------------------------
# Define mitigation outcome for each action
# ---------------------------------------------------------------------------

MITIGATION_OUTCOME = {
    "NO_ACTION": "NO_ACTION_REQUIRED",
    "MONITOR": "MONITORING_ACTIVE",
    "INVESTIGATE": "INVESTIGATION_REQUIRED",
    "ALERT_AND_LOG": "ALERT_LOGGED",
    "RATE_LIMIT": "RATE_LIMIT_SIMULATED",
    "BLOCK_AND_RATE_LIMIT": "BLOCK_AND_RATE_LIMIT_SIMULATED",
}

# ---------------------------------------------------------------------------
# Apply mitigation outcomes
# ---------------------------------------------------------------------------

mitigation_df["Mitigation_Outcome"] = (
    mitigation_df["Mitigation_Action"]
    .map(MITIGATION_OUTCOME)
)

# ---------------------------------------------------------------------------
# Verify complete mapping
# ---------------------------------------------------------------------------

missing_outcomes = (
    mitigation_df["Mitigation_Outcome"]
    .isna()
    .sum()
)

outcome_pass = missing_outcomes == 0

print("\nOutcome mapping verification:")
print(f"Missing mitigation outcomes : {missing_outcomes}")
print(f"PASS                         : {outcome_pass}")

# ---------------------------------------------------------------------------
# Outcome distribution
# ---------------------------------------------------------------------------

print("\n" + "-" * 80)
print("MITIGATION OUTCOME DISTRIBUTION")
print("-" * 80)

outcome_counts = (
    mitigation_df["Mitigation_Outcome"]
    .value_counts()
)

for outcome, count in outcome_counts.items():
    percentage = (count / len(mitigation_df)) * 100

    print(
        f"{outcome:<32}"
        f"{count:>10,} "
        f"({percentage:>6.2f}%)"
    )

# ---------------------------------------------------------------------------
# Preview mitigation outcomes
# ---------------------------------------------------------------------------

print("\n" + "-" * 80)
print("SAMPLE MITIGATION OUTCOMES")
print("-" * 80)

display(
    mitigation_df[
        [
            "Test_Index",
            "Decision_State",
            "Security_Severity",
            "Mitigation_Action",
            "Alert_Type",
            "Simulated_Block",
            "Simulated_Rate_Limit",
            "Mitigation_Outcome",
        ]
    ].head(10)
)

# ---------------------------------------------------------------------------
# Final validation
# ---------------------------------------------------------------------------

if not outcome_pass:
    raise ValueError(
        "One or more mitigation actions could not be assigned "
        "a mitigation outcome."
    )

print("\n" + "=" * 80)
print("MITIGATION OUTCOME GENERATION : PASS")
print("=" * 80)
```
