```python
import pandas as pd
import numpy as np
import os
```

```python
dataset_path = "../dataset/raw"

csv_files = [f for f in os.listdir(dataset_path) if f.endswith(".csv")]

print(f"Total CSV files: {len(csv_files)}")
print("\n".join(sorted(csv_files)))
```

```python
# Display all CSV files
for i, file in enumerate(sorted(csv_files), start=1):
    print(f"{i}. {file}")
```

```python
import os

for file in sorted(csv_files):
    path = os.path.join(dataset_path, file)
    size_mb = os.path.getsize(path) / (1024 * 1024)
    print(f"{file:<20} {size_mb:.2f} MB")
```

```python
sample_file = os.path.join(dataset_path, csv_files[0])

df = pd.read_csv(sample_file, low_memory=False)

print(f"File: {csv_files[0]}")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

df.head()
```

```python
column_sets = {}

for file in csv_files:
    path = os.path.join(dataset_path, file)
    temp_df = pd.read_csv(path, nrows=0, low_memory=False)
    temp_df.columns = temp_df.columns.str.strip()
    column_sets[file] = list(temp_df.columns)

reference = column_sets[csv_files[0]]

all_same = True

for file, cols in column_sets.items():
    if cols != reference:
        print(f"{file} has different columns.")
        all_same = False

if all_same:
    print("✅ All CSV files have the same columns.")
```

```python
for file in sorted(csv_files):
    path = os.path.join(dataset_path, file)
    temp_df = pd.read_csv(path, usecols=[" Label"], low_memory=False)
    temp_df.columns = temp_df.columns.str.strip()

    labels = temp_df["Label"].unique()
    print(f"{file}: {labels}")
```

```python
for file in sorted(csv_files):
    path = os.path.join(dataset_path, file)
    rows = sum(1 for _ in open(path)) - 1  # exclude header
    print(f"{file:<20} {rows:,} rows")
```

```python
for file in sorted(csv_files):
    path = os.path.join(dataset_path, file)
    temp_df = pd.read_csv(path, nrows=100, low_memory=False)
    temp_df.columns = temp_df.columns.str.strip()

    print(f"\n{file}")
    print(temp_df.dtypes.value_counts())
```
